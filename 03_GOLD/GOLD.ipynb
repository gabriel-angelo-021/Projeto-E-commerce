# 35. Consolidação dos indicadores de UTI

 reunindo os principais indicadores de UTI calculados durante a análise em uma única estrutura.

```python
indicadores_uti = pd.DataFrame({
    "TIPO_UTI": [
        "Adulto",
        "Pediátrica",
        "Neonatal",
        "Queimados",
        "Coronariana"
    ],
    "LEITOS_EXISTENTES": [
        hospi_gold["UTI_ADULTO_EXIST"].sum(),
        hospi_gold["UTI_PEDIATRICO_EXIST"].sum(),
        hospi_gold["UTI_NEONATAL_EXIST"].sum(),
        hospi_gold["UTI_QUEIMADO_EXIST"].sum(),
        hospi_gold["UTI_CORONARIANA_EXIST"].sum()
    ],
    "LEITOS_SUS": [
        hospi_gold["UTI_ADULTO_SUS"].sum(),
        hospi_gold["UTI_PEDIATRICO_SUS"].sum(),
        hospi_gold["UTI_NEONATAL_SUS"].sum(),
        hospi_gold["UTI_QUEIMADO_SUS"].sum(),
        hospi_gold["UTI_CORONARIANA_SUS"].sum()
    ]
})

indicadores_uti
```

# 36. Percentual de leitos SUS por tipo de UTI

 calculando o percentual de leitos destinados ao SUS para cada tipo de UTI.

```python
indicadores_uti['PERCENTUAL_SUS']=(
    indicadores_uti['LEITOS_SUS'] / indicadores_uti['LEITOS_EXISTENTES']
)* 100

indicadores_uti['PERCENTUAL_SUS'] = (
    indicadores_uti['PERCENTUAL_SUS'].round(2)
)

indicadores_uti
```

# 37. identificando UTI com maior participação do SUS

```python
maior_participacao_sus = indicadores_uti.loc[
    indicadores_uti['PERCENTUAL_SUS'].idxmax()
]

maior_participacao_sus
```

# 38. UTI com maior quantidade de leitos

```python
maior_quantidade_leitos_SUS = indicadores_uti.loc[
    indicadores_uti['LEITOS_SUS'].idxmax()
]

maior_quantidade_leitos_SUS
```

# 41. Ordenação das UTIs pelo percentual de leitos SUS

**O que vamos fazer:** ordenar os tipos de UTI do maior para o menor percentual de leitos destinados ao SUS.

**Para que serve:** facilitar a identificação das categorias com maior e menor participação proporcional do SUS.
